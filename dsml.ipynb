```python
limit=int(input('Enter number :'))
for num in range (2,limit+1):
  is_divisible=False
  k=2
  while k<=num//2:
    if num % k==0:
      is_divisible=True
      break;
    k+=1
  if not is_divisible:
    print(num,end=' ')
```

```python
#prime in given range
def isPrime(number):

    for factor in range(2, (number//2)+1):
        if number%factor == 0:

              return False

    return True

number = int(input('Enter the number '))
print(isPrime(number))
```

```python
def printPrimes(llimit, ulimit):
  for num in range(llimit,ulimit+1):
    if isPrime(num)==True:
      print(num,end=' ')
printPrimes(5,30)
```

```python
# swap 2 numbers
def swap(x,y):
  t=x
  x=y
  y=t
  return x,y
a=8
b=7
s=swap(a,b)
print(s)
```
